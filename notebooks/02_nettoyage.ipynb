#02_ nottoyage des donne Spotify 

objectif : identifier les probleme du jeu de donne justifie les traitement puis enregestrer un fichier ettoye pour analyses 

```python
from pathlib import Path
import pandas as pd

chemin_csv = Path("../data/raw/spotify-albums_data_2023.csv")

df = pd.read_csv(chemin_csv)

print("Dimension initiale :", df.shape)
display(df.head())
```

```python
colonnes_vides = df.columns[df.isna().all()].tolist()
print("Colonnes entiere vide : " , colonnes_vides)

df = df.drop(columns=colonnes_vides)

print("Dimensions apres suppression : " , df.shape)
```

```python
nb_invalides = (df["duration_ms"] <= 0).sum()

print("Durées invalides avant nettoyage :", nb_invalides)
```

```python
df = df[df["duration_ms"] > 0].copy()

print("dimension apres suppression : " , df.shape)
```

```python

df["release_date"] = pd.to_datetime(
    df["release_date"],
    errors="coerce",
    utc=True,
    format="mixed"
)

print(
    "Dates manquantes après conversion :",
    df["release_date"].isna().sum()
)
```

```python

dates_anciennes = df[df["release_date"].dt.year < 1950]

print("Nombre de morceaux avec une date avant 1950 :", len(dates_anciennes))

display(
    dates_anciennes[
        ["track_name", "artist_0", "album_name", "release_date"]
    ].head(20)
)
```

```python
print("Lignes entièrement dupliquées :", df.duplicated().sum())

print(
    "Identifiants track_id répétés :",
    df["track_id"].duplicated().sum()
)
```

```python

dossier_processed = Path("../data/processed")
dossier_processed.mkdir(parents=True, exist_ok=True)


chemin_sortie = dossier_processed / "spotify_albums_clean.csv"

df.to_csv(chemin_sortie, index=False)

print("Fichier enregistré :", chemin_sortie)
```

```python
print("Dimensions finales :", df.shape)

print("Durées invalides restantes :", (df["duration_ms"] <= 0).sum())

print("Lignes entièrement dupliquées restantes :", df.duplicated().sum())

print("Dates manquantes :", df["release_date"].isna().sum())
```

```python

df_clean = df.copy()

df_clean = df_clean.dropna(subset=["release_date"])


print("Dimensions finales :", df_clean.shape)
print("Dates manquantes :", df_clean["release_date"].isna().sum())
```

**Traitement des dates de sortie manquantes**

Lors de l'inspection des données, j'ai identifié 23 lignes dont la date de sortie (`release_date`) était manquante.

J'ai choisi de supprimer ces lignes, car la date de sortie est nécessaire pour réaliser mon analyse temporelle des albums, notamment pour étudier leur évolution au fil des années.

Cette suppression représente environ 0,005 % du jeu de données initial nettoyé (438 971 lignes). Elle a donc un impact très faible sur le volume de données disponible.

Je n'ai pas remplacé les dates manquantes par des valeurs estimées afin de ne pas introduire de dates potentiellement incorrectes dans mon analyse.
