```python
DATA_PATH = Path("../data/raw/spotify-albums_data_2023.csv")
df = pd.read_csv(DATA_PATH, low_memory=False)

print("Dataset charger avec sucees !!")
print("dimension : " , df.shape )
```

```python
display(df.describe(include="all").T)
```

```python

manquants = pd.DataFrame({
    "nombre_manquant": df.isna().sum(),
    "pourcentage_manquant": (df.isna().mean() * 100).round(2)
})

manquants = manquants.sort_values(
    "pourcentage_manquant",
    ascending=False
)

display(manquants[manquants["nombre_manquant"] > 0])
```

```python
nb_doublons = df.duplicated().sum()

print(" nombre de ligne dupliquee : " , nb_doublons)
print("pourcentage de doublons : " , round(nb_doublons / len(df)* 100 , 2),  "%")
```

```python

if "album_type" in df.columns:
    print("Nombre d'albums par type :")
    display(df["album_type"].value_counts(dropna=False))

    print("\nRépartition en pourcentage :")
    display(
        (df["album_type"].value_counts(
            normalize=True, dropna=False
        ) * 100).round(2)
    )
else:
    print("La colonne album_type n'existe pas.")
```

```python

if "release_date" in df.columns:
    dates = pd.to_datetime(df["release_date"], errors="coerce")

    print("Dates non interprétables :", dates.isna().sum())
    print("Date la plus ancienne :", dates.min())
    print("Date la plus récente :", dates.max())
else:
    print("La colonne release_date n'existe pas.")
```

```python

dates = pd.to_datetime(df["release_date"], errors="coerce")

print("Nombre de dates invalides ou manquantes :", dates.isna().sum())
print("Date la plus ancienne :", dates.min())
print("Date la plus récente :", dates.max())

print("\nExemples de dates originales :")
display(df["release_date"].dropna().head(10))
```

```python

df_dates = df.copy()
df_dates["release_date_convertie"] = pd.to_datetime(
    df_dates["release_date"], errors="coerce", utc=True
)

display(
    df_dates.loc[
        df_dates["release_date_convertie"].dt.year < 1950,
        ["album_name", "track_name", "release_date", "album_type"]
    ].sort_values("release_date").head(20)
)
```

```python

print("Lignes entièrement dupliquées :", df.duplicated().sum())


print("track_id répétés :", df["track_id"].duplicated().sum())


print("Durées nulles ou négatives :", (df["duration_ms"] <= 0).sum())

print("\nLes 10 morceaux les plus longs :")
display(
    df.nlargest(10, "duration_ms")[
        ["track_name", "artist_0", "duration_ms"]
    ]
)
```

```python

doublons_track_id = df[df["track_id"].duplicated(keep=False)]

display(
    doublons_track_id[
        ["track_id", "track_name", "artist_0", "album_name"]
    ].sort_values("track_id")
)
```

```python

durees_invalides = df[df["duration_ms"] <= 0]

display(
    durees_invalides[
        ["track_name", "artist_0", "duration_ms"]
    ]
)
```
